````python
import cirq
import matplotlib.pyplot as plt
````

````python
q=cirq.GridQubit(0,0)
circuit = cirq.Circuit(
    cirq.H(q),
    cirq.measure(q,key='m')
)
````

````python
simulator = cirq.Simulator()
result=simulator.run(circuit,repetitions=5000)
counts = result.histogram(key='m')
````

````python
print("Comptes :",counts)
probabilities = {k: v/5000 for k, v in counts.items()}
print("Probabilités :", probabilities)
````

Effet des portes

````python
qubit=cirq.GridQubit(0,0)
sim = cirq.Simulator()

gates={"H": cirq.H(q),
    "X": cirq.X(q),
    "Y": cirq.Y(q),
    "Z": cirq.Z(q)}

for name,gate in gates.items():
  circuit =cirq.Circuit(gate)
  result=sim.simulate(circuit)
  print(name, "→", result.final_state_vector)
circuit =cirq.Circuit(cirq.H(qubit),cirq.H(qubit))

print("H puis H →", sim.simulate(circuit).final_state_vector)
````



```
# Ce texte est au format code
| Porte | Effet            |    |
| ----- | ---------------- | -- |
| H     | superposition    |    |
| X     | inverse 0 ↔ 1    |    |
| Y     | rotation + phase |    |
| Z     | change phase de  | 1⟩ |
| H→H   | revient à        | 0⟩ |

```

````python
import cirq, numpy as np

q = cirq.GridQubit(0,0)

alpha = 2+1j
beta = 1-1j
norm = np.sqrt(abs(alpha)**2 + abs(beta)**2)
alpha /= norm
beta /= norm

state = np.array([alpha, beta], dtype=complex)

sim = cirq.Simulator()

def run_gate(gate):
    circuit = cirq.Circuit(
        cirq.StatePreparationChannel(state)(q),
        gate(q)
    )
    result = sim.simulate(circuit)
    print(gate, "→", result.final_state_vector)

run_gate(cirq.X)
run_gate(cirq.Z)
run_gate(cirq.H)
````

Intrication

````python
import cirq

q0, q1 = cirq.GridQubit.rect(1,2)

circuit = cirq.Circuit(
    cirq.H(q0),
    cirq.CNOT(q0, q1),
    cirq.measure(q0, q1, key='m')
)

sim = cirq.Simulator()
result = sim.run(circuit, repetitions=1000)
counts = result.histogram(key='m')

print(counts)
````
